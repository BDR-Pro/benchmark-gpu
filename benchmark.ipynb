```python
import time
import torch

print(f"PyTorch Version: {torch.__version__}")
print(f"CUDA Available: {torch.cuda.is_available()}")

if torch.cuda.is_available():
    print(f"Found GPU(s): {torch.cuda.get_device_name(0)}")
    try:
        # PyTorch handles memory growth dynamically by default.
        device = torch.device("cuda:0")

        print("Generating dummy GPU load...")
        matrix_size = 8192  # Adjust this size for more or less load
        a = torch.randn(matrix_size, matrix_size, device=device)
        b = torch.randn(matrix_size, matrix_size, device=device)

        num_iterations = 100000000

        # Warm-up run (helps stabilize CUDA initial overhead)
        _ = torch.matmul(a, b)
        torch.cuda.synchronize()

        start_time = time.time()
        for _ in range(num_iterations):
            c = torch.matmul(a, b)

        # Ensure GPU computations complete before stopping the timer
        torch.cuda.synchronize()
        end_time = time.time()

        total_time = end_time - start_time
        print(f"Completed {num_iterations} matrix multiplications on GPU.")
        print(f"Total time taken: {total_time:.4f} seconds")
        print(
            "Average time per multiplication:"
            f" {total_time / num_iterations:.4f} seconds"
        )

    except Exception as e:
        print(f"Error initializing GPU: {e}")
        print("Falling back to CPU if GPU cannot be used.")

        # Fallback to CPU if GPU fails
        device = torch.device("cpu")
        print("Generating dummy CPU load (GPU error fallback)...")
        matrix_size = 2048  # Smaller size for CPU
        a = torch.randn(matrix_size, matrix_size, device=device)
        b = torch.randn(matrix_size, matrix_size, device=device)

        num_iterations = 5
        start_time = time.time()
        for _ in range(num_iterations):
            c = torch.matmul(a, b)
        end_time = time.time()

        total_time = end_time - start_time
        print(f"Completed {num_iterations} matrix multiplications on CPU.")
        print(f"Total time taken: {total_time:.4f} seconds")
        print(
            "Average time per multiplication:"
            f" {total_time / num_iterations:.4f} seconds"
        )

else:
    print("No GPU found. Performing dummy load on CPU.")
    device = torch.device("cpu")
    print("Generating dummy CPU load...")

    matrix_size = 2048  # Adjust this size for more or less load on CPU
    a = torch.randn(matrix_size, matrix_size, device=device)
    b = torch.randn(matrix_size, matrix_size, device=device)

    num_iterations = 5
    start_time = time.time()
    for _ in range(num_iterations):
        c = torch.matmul(a, b)
    end_time = time.time()

    total_time = end_time - start_time
    print(f"Completed {num_iterations} matrix multiplications on CPU.")
    print(f"Total time taken: {total_time:.4f} seconds")
    print(
        "Average time per multiplication:"
        f" {total_time / num_iterations:.4f} seconds"
    )

print("Dummy load and benchmark complete.")
```
